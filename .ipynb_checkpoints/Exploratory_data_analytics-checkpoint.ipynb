# Feature Engineering 

```python
df['signup_date'] = pd.to_datetime(df['signup_date'])
```

```python
df['signup_year'] = df['signup_date'].dt.year
df['signup_month'] = df['signup_date'].dt.month
df['signup_day'] = df['signup_date'].dt.day
```

```python
df.drop('signup_date', axis=1, inplace=True)
```

# Relationship Between Features and Target Variable

## Numerical Features vs Churn

```python
# Monthly fee vs Churn :
sns.boxplot(x='churn', y='monthly_fee', data=df)
plt.title("Monthly Fee vs Churn")
plt.show()
```

### Observation

- Monthly fee shows some variation between churned and non-churned customers.
- Further analysis using machine learning can help determine its importance in predicting churn.

```python
# Payment failures vs churn 
sns.boxplot(x='churn', y='payment_failures', data=df)
plt.title("payment fialures vs churn")
plt.show()
```

### Observation

- Payment failures show variation between churned and non-churned customers.
- Customers with more payment issues may be more likely to leave the service.

```python
# support tickets vs churn 
sns.boxplot(x='churn', y='support_tickets', data=df)
plt.title("support tickets vs Churn")
plt.show()
```

### Observation

- Customers with different numbers of support tickets exhibit different churn behavior.
- Customer support interactions may have an impact on customer retention.

```python
# last login days ago vs churn 
sns.boxplot(x='churn', y='last_login_days_ago', data=df)
plt.title("last login days ago vs Churn")
plt.show()
```

### Observation

- Customer inactivity differs between churned and retained customers.
- Customers who have not logged in for a longer time may be more likely to churn.

```python
# Tenure Months vs churn 
sns.boxplot(x='churn', y='tenure_months', data=df)
plt.title("Tenure months  vs Churn")
plt.show()
```

### Observation

- Customer tenure appears to be related to churn.
- Long-term customers may show different churn behavior compared to newer customers.

# Categorical Features vs Churn 

```python
#plan type vs churn 
plt.figure(figsize=(6,4))

sns.countplot(
    x='plan_type',
    hue='churn',
    data=df
)

plt.title("Plan Type vs Churn")
plt.xlabel("Plan Type")
plt.ylabel("Count")
plt.show()
```

```python
plan_churn = pd.crosstab(
    df['plan_type'],
    df['churn'],
    normalize='index'
)

plan_churn.plot(kind='bar', figsize=(6,4))

plt.title("Churn Rate by Plan Type")
plt.ylabel("Proportion")
plt.legend(title="Churn")
plt.show()
```

### Observation

- Churn distribution varies across different subscription plans.
- Some plan types appear to have a higher number of churned customers than others.
- This suggests that subscription plan may influence customer retention.

```python
plt.figure(figsize=(10,4))

sns.countplot(
    x='signup_month',
    hue='churn',
    data=df
)

plt.title("Signup Month vs Churn")
plt.xlabel("Signup Month")
plt.ylabel("Count")
plt.show()
```

### Observation

- Customers joined the service throughout different months.
- Churn counts vary across signup months.
- Seasonal or promotional factors may influence customer retention.

```python
plt.figure(figsize=(6,4))

sns.countplot(
    x='signup_year',
    hue='churn',
    data=df
)

plt.title("Signup Year vs Churn")
plt.xlabel("Signup Year")
plt.ylabel("Count")
plt.show()
```

### Observation

- Customers who joined in different years show varying churn distributions.
- This may indicate changes in customer behavior or business strategies over time.

### Correlation Analysis

```python
df_numeric = df.drop(columns=['user_id'])
corr = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title("Correlation Heatmap of Numerical Features")
plt.show()
```

### Observation

- The correlation heatmap shows that most numerical features have very weak correlations with one another.
- The correlation coefficients range approximately from **-0.06 to 0.03**, indicating little or no linear relationship between the features.
- No pairs of numerical features exhibit strong positive or negative correlation.
- This suggests that the features provide different information and there are no significant multicollinearity issues.
- The `user_id` column is only an identifier and does not provide meaningful information for correlation analysis.

### Key Business Insights 


- The dataset is clean, containing no missing values or duplicate records, making it suitable for analysis and machine learning.

- Most numerical features do not contain significant outliers, indicating consistent data quality.

- Customer churn varies across different numerical features such as monthly fee, tenure, support tickets, payment failures, and last login days, suggesting that these features may influence customer retention.

- Different subscription plans show different churn distributions, indicating that plan type may play a role in customer churn.

- Customer signup patterns vary across different months and years, which may help identify customer acquisition trends.

- The correlation analysis shows very weak relationships among the numerical features, indicating that the features provide distinct information and do not suffer from significant multicollinearity.

- Based on the exploratory analysis, features such as tenure, payment failures, support tickets, monthly fee, last login days, and plan type can be considered important candidates for building customer churn prediction models.

# Conclusion

This exploratory data analysis provided a better understanding of customer characteristics and churn patterns. The dataset was found to be clean and well-structured, requiring minimal preprocessing. Several numerical and categorical features showed differences in churn behavior, while the correlation analysis indicated low multicollinearity among numerical variables. These findings provide a strong foundation for the next stage of the project, where machine learning models will be developed to predict customer churn.
